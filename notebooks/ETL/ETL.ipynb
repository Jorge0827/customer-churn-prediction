```python
import pandas as pd
```

```python
df = pd.read_csv("../data/WA_Fn-UseC_-Telco-Customer-Churn.csv")
```

```python
df.info()
```

```python
df["TotalCharges"].unique()
```

Convertir TotalCharges a numero

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")
```

Revisar si hay nulos

```python
df["TotalCharges"].isnull().sum()
```

Esos 11 son clientes nuevos que hay no llevan tiempo en el servicio

```python
df["TotalCharges"].fillna(0, inplace=True)
```

```python
df.info()
```

Revisar si existen duplicados

```python
df.duplicated().sum()
```

Se realizó una verificación de registros duplicados en el dataset utilizando la función duplicated(). El resultado muestra que no existen registros duplicados, lo que indica que cada observación representa un cliente único dentro del conjunto de datos. Por lo tanto, no fue necesario aplicar procesos de eliminación de duplicados durante la etapa de limpieza de datos.

revisar si hay variables categoricas que deben convertirse a numerico ----------->>>

```python
df.select_dtypes(include="object").columns
```

Ahora debemos convertir Churn a numerico, 0 y 1 porque nuestro modelo no entenderá valores numericos

```python
df["Churn"] = df["Churn"].map({"No": 0, "Yes": 1})
```

```python
df["Churn"].value_counts()
```

Eliminar también CustomerID

```python
df = df.drop("customerID", axis=1)
```

Convertir todas las variables categoricas a numericas

```python
df = pd.get_dummies(df, drop_first=True)
```
